```python
corpus = [
    "<s> cô ấy dạy môn tin học </s>",
    "<s> anh dạy môn toán </s>",
    "<s> cô ấy học toán anh ấy dạy </s>",
    "<s> môn toán môn tin đều hay </s>",
    "<s> anh ấy dạy môn toán hay môn tin </s>"
]
```

```python
from collections import Counter, defaultdict

# Tách token cho từng câu
tokenized_corpus = [sentence.split() for sentence in corpus]
```

```python
unigram_counts = Counter()

for sentence in tokenized_corpus:
    for token in sentence:
        if token not in ["<s>", "</s>"]:
            unigram_counts[token] += 1

total_unigrams = sum(unigram_counts.values())
```

```python
unigram_probs = {}

print("UNIGRAM MODEL")
print("Total tokens:", total_unigrams)
print()
for word, count in unigram_counts.items():
    unigram_probs[word] = count / total_unigrams
    print(f"P({word}) = {unigram_probs[word]:.4f}")
```

```python
bigram_counts = Counter()
unigram_context_counts = Counter()

for sentence in tokenized_corpus:
    for i in range(len(sentence) - 1):
        w_prev = sentence[i]
        w_curr = sentence[i + 1]

        bigram_counts[(w_prev, w_curr)] += 1
        unigram_context_counts[w_prev] += 1
```

```python
bigram_probs = {}

print("BIGRAM MODEL")
print("Total tokens:", sum(bigram_counts.values()))
print()
for (w_prev, w_curr), count in bigram_counts.items():
    bigram_probs[(w_prev, w_curr)] = count / unigram_context_counts[w_prev]
    print(f"P({w_curr} | {w_prev}) = {bigram_probs[(w_prev, w_curr)]: .4f}")
```

```python
def bigram_probability(w_prev, w_curr):
    return bigram_probs.get((w_prev, w_curr), 0.0)

print("EXAMPLES")
print("P(ấy | cô) =", bigram_probability("cô", "ấy"))
print("P(môn | dạy) =", bigram_probability("dạy", "môn"))
print("P(tin | môn) =", bigram_probability("môn", "tin"))
print("P(toán | môn) =", bigram_probability("môn", "toán"))
```
